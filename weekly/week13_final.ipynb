# Final round - BBO capstone lock-in

The last submission of Stage 2. Seven functions lock their best-ever coordinates - guaranteed to reproduce exactly, per the determinism established by the week 3 and week 10 reproduction tests (identical returns to 12-15 decimal places).

**The one deliberate exception: f1.** Three equal steps along one line returned -0.0053 -> +0.0027 -> +0.0315, roughly a tenfold increase per step with no sign of the growth turning over. Rather than banking the guaranteed 0.0315, the final submission takes one further step along the same line to (0.44, 0.45). Decision taken knowingly: the upside if the pattern holds is another order of magnitude; the downside is returning less than 0.0315 with no round left to recover.


```python
import os, sys, warnings
warnings.filterwarnings("ignore")
_root = os.getcwd()
while not os.path.exists(os.path.join(_root, "bo.py")) and os.path.dirname(_root) != _root:
    _root = os.path.dirname(_root)
os.chdir(_root); sys.path.insert(0, _root)
import numpy as np
import bo

finals = {}
for fid in bo.FUNC_IDS:
    X, y, d = bo.load(fid)
    i = int(np.argmax(y))
    finals[fid] = (X[i], float(y[i]))

# f1: one further step along the signal line instead of banking 0.0315.
finals[1] = (np.array([0.440000, 0.450000]), None)

for fid in bo.FUNC_IDS:
    sub = bo.format_submission(finals[fid][0])
    bo.validate_submission(sub, len(finals[fid][0]))
    g = f"guaranteed {finals[fid][1]:.6g}" if finals[fid][1] is not None else "signal push"
    print(f"f{fid}:  {sub}   ({g})")
```
